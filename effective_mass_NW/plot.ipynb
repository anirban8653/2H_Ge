```python
import numpy as np
import matplotlib.pyplot as plt

data  = np.load("eff_data.npy")
L = data[:,0]
eff = data[:,1]

plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = 'DejaVu Serif'

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(L, eff, marker='o', linestyle='-', color='#1f77b4',
        markersize=6, linewidth=1.8, markeredgecolor='blue', markeredgewidth=0.5, markerfacecolor = "w")

ax.set_xlabel('L', fontsize=20)
ax.set_ylabel(r'$m_x^* (m_e)$', fontsize=20)

# Ticks pointing inward, on all sides
ax.tick_params(axis='both', which='major', labelsize=22,
                direction='in', length=6, width=1.2, top=True, right=True)
ax.tick_params(axis='both', which='minor', direction='in',
                length=3, width=1, top=True, right=True)

# Slightly thicker spines to match tick weight
for spine in ax.spines.values():
    spine.set_linewidth(1.2)

# ax.minorticks_on()

plt.tight_layout()
plt.savefig("Effective_mass_modulation.pdf")
plt.savefig("Effective_mass_modulation.png", dpi=300)
plt.show()
```
