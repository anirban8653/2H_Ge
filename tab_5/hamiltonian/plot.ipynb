```python
import numpy as np
import matplotlib.pyplot as plt

# --- Data and Parameters ---
kx_length = 51
num_eigvals = 24
N = 100


plt.rcParams.update({'figure.autolayout': True})
plt.figure(figsize=(3.7, 5))
plt.rcParams["font.family"] = "DejaVu Serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"

data = np.loadtxt(f"band_kx_N{N}_L300.dat")

cmap = plt.cm.Blues
colors = cmap([0.85, 0.65, 0.45])


for i in range(1, num_eigvals + 1):
    plt.plot(data[:, 0], data[:, i] * 1000,color = "k")



plt.xlabel(r"$k_x (\AA^{-1})$", fontsize=18)
plt.ylabel("Energy (meV)", fontsize=18)
# plt.axhline(y=0, ls='--', color='k')
plt.axvline(x=0, ls='--', color='k')
plt.ylim(-31, 0)
plt.xlim(-0.02, 0.02)

plt.tick_params(direction='in', length=7, width=1., colors='k',
                    bottom=True, top=True, left=True, right=True, labelsize=18)

plt.tight_layout()
plt.savefig("valance_band_nw.png", dpi=300)
plt.savefig("valance_band_nw.pdf")
plt.savefig("valance_band_nw.svg")
plt.tight_layout()
plt.show() 
```
