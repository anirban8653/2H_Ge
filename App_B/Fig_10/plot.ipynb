```python
import numpy as np
import matplotlib.pyplot as plt

data  = np.load("eff_data.npy")
L = data[:,0]
eff = data[:,1]
plt.rcParams.update({
    "figure.figsize": [4.5, 4],
    "font.family": "Serif",
    "mathtext.fontset": "dejavuserif",
    "axes.linewidth": 0.8
})

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(L, eff, marker='o', linestyle='-', color='#1f77b5', linewidth=1.8, markeredgecolor='#1f77b5', 
        markeredgewidth=0.5, markerfacecolor = "#1f77b5")

ax.set_xlabel('$L$ (nm)', fontsize = 16)
ax.set_ylabel(r'$m_x^* (m_0)$', fontsize = 16)

# Ticks pointing inward, on all sides
ax.tick_params(axis='both', which='major', labelsize=16,
                direction='in', length=6, width=1.2, top=True, right=True)
ax.tick_params(axis='both', which='minor', direction='in',
                length=3, width=1, top=True, right=True)

# Slightly thicker spines to match tick weight
for spine in ax.spines.values():
    spine.set_linewidth(1.2)
# ax.grid()

# ax.minorticks_on()

plt.tight_layout()
plt.savefig("mx_variation.pdf")
plt.savefig("mx_variation.png", dpi=300)
plt.show()
```
